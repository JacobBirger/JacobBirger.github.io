```python
import pandas as pd
import altair as alt

url = 'https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_data/main/building_inventory.csv'
df = pd.read_csv(url)

df_clean = df[df['Year Acquired'] > 1700].copy()
df_grouped = df_clean.groupby('Bldg Status', as_index=False)['Square Footage'].sum()

plot1 = alt.Chart(df_grouped).mark_bar().encode(
    x=alt.X('Bldg Status:N', title='Building Status'),
    y=alt.Y('Square Footage:Q', title='Total Square Footage'),
    color=alt.Color('Bldg Status:N', scale=alt.Scale(scheme='set2'), legend=None)
).properties(
    width=400,
    title='Total Square Footage by Building Status'
)

plot1.save('plot1.json')
brush = alt.selection_interval()

plot2 = alt.Chart(url).mark_circle(size=50).transform_filter(
    alt.datum['Year Acquired'] > 1700
).encode(
    x=alt.X('Year Acquired:Q', scale=alt.Scale(domain=[1750, 2020])),
    y=alt.Y('Square Footage:Q'),
    color=alt.condition(brush, 'Bldg Status:N', alt.value('lightgray'), scale=alt.Scale(scheme='set2')),
    tooltip=['Agency Name:N', 'Year Acquired:Q', 'Square Footage:Q']
).properties(
    width=500,
    title='Building Square Footage over Time'
).add_params(
    brush
)

plot2.save('plot2.json')
```
